```python
import time
from collections import deque

# -------------------------------
# Graph Implementation
# DFS and BFS
# -------------------------------

# User input for number of vertices
n = int(input("Enter number of vertices: "))

# Create graph using adjacency list
graph = [[] for _ in range(n)]

# User input for number of edges
e = int(input("Enter number of edges: "))

print("Enter edges (source destination):")

for i in range(e):
    u, v = map(int, input().split())

    # Undirected graph
    graph[u].append(v)
    graph[v].append(u)

# Display graph
print("\nGraph (Adjacency List):")
for i in range(n):
    print(i, "->", graph[i])


# -------------------------------
# DFS
# -------------------------------

def dfs(graph, start):
    visited = [False] * len(graph)
    result = []

    def dfs_recursive(vertex):
        visited[vertex] = True
        result.append(vertex)

        for neighbour in graph[vertex]:
            if not visited[neighbour]:
                dfs_recursive(neighbour)

    dfs_recursive(start)
    return result


# -------------------------------
# BFS
# -------------------------------

def bfs(graph, start):
    visited = [False] * len(graph)
    queue = deque()
    result = []

    visited[start] = True
    queue.append(start)

    while queue:
        vertex = queue.popleft()
        result.append(vertex)

        for neighbour in graph[vertex]:
            if not visited[neighbour]:
                visited[neighbour] = True
                queue.append(neighbour)

    return result


# -------------------------------
# User input for starting vertex
# -------------------------------

start = int(input("\nEnter starting vertex: "))


# -------------------------------
# DFS Execution Time
# -------------------------------

start_time = time.perf_counter()

dfs_result = dfs(graph, start)

end_time = time.perf_counter()

dfs_time = end_time - start_time


# -------------------------------
# BFS Execution Time
# -------------------------------

start_time = time.perf_counter()

bfs_result = bfs(graph, start)

end_time = time.perf_counter()

bfs_time = end_time - start_time


# -------------------------------
# Output
# -------------------------------

print("\nDFS Traversal:")
print(" -> ".join(map(str, dfs_result)))

print("\nBFS Traversal:")
print(" -> ".join(map(str, bfs_result)))

print("\nExecution Time:")
print("DFS:", dfs_time, "seconds")
print("BFS:", bfs_time, "seconds")


# -------------------------------
# Time Complexity
# -------------------------------

print("\nTime Complexity:")
print("DFS: O(V + E)")
print("BFS: O(V + E)")
```
