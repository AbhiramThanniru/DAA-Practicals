```python
import time

# Prim's Algorithm

# Adjacency matrix
graph = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]

vertices = len(graph)

# Start execution timer
start_time = time.perf_counter()

# To store vertices included in MST
selected = [False] * vertices

# Start from vertex 0
selected[0] = True

mst_edges = []
total_cost = 0

# MST contains vertices - 1 edges
for _ in range(vertices - 1):

    minimum = float('inf')
    x = -1
    y = -1

    for i in range(vertices):
        if selected[i]:
            for j in range(vertices):
                if not selected[j] and graph[i][j] != 0:
                    if graph[i][j] < minimum:
                        minimum = graph[i][j]
                        x = i
                        y = j

    selected[y] = True
    mst_edges.append((x, y, minimum))
    total_cost += minimum

# End execution timer
end_time = time.perf_counter()

execution_time = end_time - start_time

# Display MST
print("Edges in Minimum Spanning Tree:")

for u, v, weight in mst_edges:
    print(f"{u} - {v} : {weight}")

print("\nTotal cost of MST:", total_cost)

print("\nExecution Time:", execution_time, "seconds")

print("\nTime Complexity: O(V^2)")
```
