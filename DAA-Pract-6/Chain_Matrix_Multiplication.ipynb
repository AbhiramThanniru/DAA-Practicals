```python
import time

def matrix_chain_order(p):
    n = len(p) - 1
    if n <= 0:
        return 0
        
    dp = [[0] * n for _ in range(n)]
 
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            dp[i][j] = float('inf')
            for k in range(i, j):
                cost = (
                    dp[i][k] + dp[k + 1][j] + p[i] * p[k + 1] * p[j + 1]
                )
                dp[i][j] = min(dp[i][j], cost)
                
    return dp[0][n - 1]

if __name__ == "__main__":
    user_input = input("Enter the dimensions (e.g., 10 20 30 40): ")
    p = [int(x) for x in user_input.replace(',', ' ').split()]

    if len(p) < 2:
        print("Error: You must enter at least two dimensions (e.g., 10 20).")
    else:
        start = time.time()
        result = matrix_chain_order(p)
        end = time.time()

        print(f"\nMinimum number of multiplications: {result}")
        print(f"Execution time: {end - start:.6f} seconds")
        print("\n-----Complexity-----")
        print("Best case Time Complexity: O(n^3)")
        print("Average case Time Complexity: O(n^3)")
        print("Worst case Time Complexity: O(n^3)")
        print("Space Complexity: O(n^2)")

```
